```python
import pandas as pd

trans = pd.read_csv(r"D:\Project\AuditTrace\agentic-ai-fraud-detection\data\train_transaction.csv")
identity = pd.read_csv(r"D:\Project\AuditTrace\agentic-ai-fraud-detection\data\train_identity.csv")

print(trans.shape, identity.shape)
```

```python
#Merge transaction + identity
df = trans.merge(identity, on="TransactionID", how="left")
df.head()
```

```python
#Understand Time - Convert to usable time
df["TransactionDT"] = pd.to_datetime(df["TransactionDT"], unit="s")
df = df.sort_values("TransactionDT")
```

```python
#Create synthetic customer ID - enabling per-customer time-series modeling
df["customer_id"] = (
    df["card1"].astype(str) + "_" +
    df["card2"].astype(str)
)
```

```python
#First Fraud-over-Time Check
df.set_index("TransactionDT")["isFraud"].resample("D").mean().plot(
    title="Fraud Rate Over Time"
)
```

```python
#Select Core Columns
core_cols = [
    "TransactionDT",
    "TransactionAmt",
    "customer_id",
    "card1",
    "card2",
    "ProductCD",
    "isFraud"
]

df_core = df[core_cols].dropna()
df_core.head()
```

```python
#Save Clean Base Dataset
df_core.to_csv(r"D:\Project\AuditTrace\agentic-ai-fraud-detection\data\base_transactions.csv", index=False)
```
